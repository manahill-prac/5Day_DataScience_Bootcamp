```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

df = pd.DataFrame({
    'Name':     ['Manahil', 'Sara', 'Ahmed', 'Zara', 'Bilal', 'Hina'],
    'DOB':      ['2005-07-03', '2001-03-22', '1995-11-08', '2000-07-30', '1980-01-15', '2002-09-10'],
    'Price':    [500, 1200, 300, 800, 1500, 400],
    'Quantity': [3, 1, 5, 2, 4, 6],
})

print(df)
```

# TYPE 1 — Extract from Dates


```python
import datetime

df['DOB'] = pd.to_datetime(df['DOB'])

current_year = datetime.datetime.now().year

df['Age'] = current_year - df['DOB'].dt.year

print(df[['Name', 'Age']])
```

# TYPE 2 — Create Age Group/Category from Age


```python
df['Age_Group'] = pd.cut(
    df['Age'],
    bins=[0, 25, 40, 60],
    labels=['Young', 'Mid-Career', 'Senior']
)

print(df[['Name', 'Age', 'Age_Group']])
```

# TYPE 3 — Combine Price and Quantity


```python
df['Total_Spend'] = df['Price'] * df['Quantity']

print(df[['Name', 'Price', 'Quantity', 'Total_Spend']])
```
