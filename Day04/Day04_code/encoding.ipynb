LabelEncoder

```python
#dataset for encoding
import pandas as pd

df = pd.DataFrame({
    'Name':   ['Ali', 'Sara', 'Ahmed', 'Zara', 'Bilal', 'Hina'],
    'Age': [18, 20, 22, 19,26,18],
    'Education': ['Matric', 'Bachelors', 'Masters', 'Bachelors', 'PhD', 'Matric'],
    'City':   ['Karachi', 'Lahore', 'Islamabad', 'Karachi', 'Lahore', 'Islamabad'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Salary': [30000, 55000, 75000, 50000, 95000, 28000]
})

print("Before Encoding:")
print(df)
```

# Label Encoding

```python
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder()
df['Education_Encoded'] = le.fit_transform(df['Education'])

print("\nAfter Label Encoding:")
print(df)
```

# Ordinal Encoding


```python
# Ordinal Encoding — we define the order ourselves
from sklearn.preprocessing import OrdinalEncoder
encoder = OrdinalEncoder(categories=[['Matric', 'Bachelors', 'Masters', 'PhD']])
df['Education_Encoded'] = encoder.fit_transform(df[['Education']])

print(df)

```

# One-Hot Encoding


```python
df_encoded = pd.get_dummies(df, columns=['City', 'Gender'], drop_first=True)

print("\nAfter One-Hot Encoding:")
print(df_encoded)



```
