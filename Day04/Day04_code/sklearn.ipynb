```python
import pandas as pd
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OrdinalEncoder, OneHotEncoder
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer

df = pd.DataFrame({
    'Name':      ['Ali', 'Sara', 'Ahmed', 'Zara', 'Bilal','Hina', 'Omar', 'Ayesha', 'Kamil', 'Nida'],
    'Age':       [22, 35, 45, 28, 40, 30, 25, 38, 50, 27],
    'Salary':    [30000, 55000, 75000, 50000, 95000,
                  28000, 35000, 62000, 88000, 42000],
    'Education': ['Matric', 'Bachelors', 'Masters', 'Bachelors','PhD', 'Matric', 'Bachelors', 'Masters', 'PhD', 'Bachelors'],
    'City':      ['Karachi', 'Lahore', 'Islamabad', 'Karachi', 'Lahore','Islamabad', 'Karachi', 'Lahore', 'Islamabad', 'Karachi'],
    'Purchased': [0, 1, 1, 0, 1, 0, 0, 1, 1, 1]
})

print(df)
```

# Separate Features & Targets

```python
x = df[['Age', 'Salary', 'Education', 'City']]
y = df['Purchased']

print("Features (X):")
print(x.to_string())

print("\nTarget (y):")
print(y.to_string())
```

# Train Test Split


```python
x_train, x_test, y_train, y_test = train_test_split(
    x, y,
    test_size=0.2,
    random_state=42
)


print(f"Training rows : {len(x_train)}")
print(f"Testing rows  : {len(x_test)}")

print(len(x_train))
print(len(x_test))
```

```python
numerical_cols   = ['Age', 'Salary']

ordinal_cols     = ['Education']

nominal_cols     = ['City']

```

# Build the ColumnTransformer


```python
preprocessor = ColumnTransformer(transformers=[
    ('scale',
     StandardScaler(),
     numerical_cols),

    ('ordinal',
     OrdinalEncoder(categories=[
         ['Matric', 'Bachelors', 'Masters', 'PhD']
     ]),
     ordinal_cols),

    ('onehot',
     OneHotEncoder(drop='first'),
     nominal_cols)

])

```

# Build the Pipeline


```python
pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
])

print("Pipeline created successfully!")
print(pipeline)

```

# Fit and Transform


```python
# Fit on training data — Pipeline LEARNS from train only
x_train_transformed = pipeline.fit_transform(x_train)

# Transform test data — Pipeline APPLIES what it learned
x_test_transformed  = pipeline.transform(x_test)

print("X_train after Pipeline:")
print(x_train_transformed.round(2))

print(f"Shape before: {x_train.shape}")
print(f"Shape after : {x_train_transformed.shape}")



```
